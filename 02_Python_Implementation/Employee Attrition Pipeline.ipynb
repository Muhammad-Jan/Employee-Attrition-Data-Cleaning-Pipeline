# Data Type Fixing

```python
df.columns = df.columns.str.strip()
```

```python
df["HourlyRate"] = df["HourlyRate"].replace(r"[\$,]", "", regex=True).astype(float)
```

```python
df["AnnualSalary"] = pd.to_numeric(df["AnnualSalary"], errors="coerce")
df["DrivingCommuterDistance"] = pd.to_numeric(df["DrivingCommuterDistance"], errors="coerce")
df["NumCompaniesPreviouslyWorked"] = pd.to_numeric(df["NumCompaniesPreviouslyWorked"], errors="coerce")
df["AnnualProfessionalDevHrs"] = pd.to_numeric(df["AnnualProfessionalDevHrs"], errors="coerce")
```

# Handling Missing Values

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df.replace("N/A", np.nan, inplace=True)

df["AnnualProfessionalDevHrs"].fillna(df["AnnualProfessionalDevHrs"].median(), inplace=True)
df["NumCompaniesPreviouslyWorked"].fillna(df["NumCompaniesPreviouslyWorked"].median(), inplace=True)
df["PaycheckMethod"].fillna(df["PaycheckMethod"].mode()[0], inplace=True)
df["TextMessageOptIn"].fillna("No", inplace=True)
```

# Removing Duplicates

```python
df.drop_duplicates(inplace=True)
```

# Fixing Inconsistent Values

```python
df['PaycheckMethod'] = df['PaycheckMethod'].replace({
    'Mail_Check': 'Mailed Check',
    'MailedCheck': 'Mailed Check',
    'Mail Check': 'Mailed Check',
    'Direct_Deposit': 'Direct Deposit',
    'DirectDeposit': 'Direct Deposit'
})

df["Gender"] = df["Gender"].replace({
    "Prefer Not to Answer": np.nan
})

df['JobRoleArea'] = df['JobRoleArea'].replace({
    'Information_Technology': 'Information Technology',
    'InformationTechnology': 'Information Technology',
    'Human_Resources': 'Human Resources',
    'HumanResources': 'Human Resources'
})
```

```python
df.isnull().sum()
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```

# Outlier Treatment (IQR Method)

```python
Q1 = df["AnnualSalary"].quantile(0.25)
Q3 = df["AnnualSalary"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[(df["AnnualSalary"] >= lower_bound) & (df["AnnualSalary"] <= upper_bound)]
```

# Feature Standardization

```python
df["JobRoleArea"] = df["JobRoleArea"].str.replace("_", " ")
df["Turnover"] = df["Turnover"].map({"Yes": 1, "No": 0})
df["TextMessageOptIn"] = df["TextMessageOptIn"].map({"Yes": 1, "No": 0})
```
